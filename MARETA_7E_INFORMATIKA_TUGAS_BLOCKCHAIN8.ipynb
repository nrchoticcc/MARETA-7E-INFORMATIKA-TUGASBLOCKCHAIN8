## SECTION 5 — HOW DEFI GENERATES RETURN · `EXPLORE`

| Mekanisme | Sumber return | Siapa yang membayar? | Risiko utama |
|---|---|---|---|
| **Lending** | Bunga (*interest*) | Peminjam | Gagal likuidasi, bug smart contract, depeg collateral |
| **Liquidity Providing** | *Trading fee* | Trader yang melakukan swap | Impermanent loss, volume turun |
| **Staking** | *Protocol/network reward* | Penerbitan token baru + fee jaringan | Slashing, lock-up, harga token turun |
| **Yield Farming** | *Incentive/reward token* | Treasury protokol / emisi token baru | Insentif berhenti, token reward anjlok, rug pull |
| **Arbitrage** | Selisih harga antar pasar | Peluang selisih harga di pasar | Sangat berisiko, butuh modal dan kecepatan; bukan praktik mahasiswa |

> **Prinsip penting:** Jika sumber uang/reward tidak dapat dijelaskan, ada kemungkinan skema tersebut tidak berkelanjutan atau merupakan scam.

### 5.1 Dari mana uang/reward berasal?

- **Emisi token baru vs fee/bunga:** fee dan bunga berasal dari aktivitas ekonomi pengguna, sedangkan emisi token baru dapat sekadar membagikan token baru tanpa pendapatan eksternal yang sebanding. ## SECTION 6 — ZERO-COST DEFI SIMULATION · `EXPERIMENT`

> 💵 **`modal_simulasi = 1000` adalah $1,000 UANG VIRTUAL, BUKAN UANG NYATA.**
> Tidak ada wallet, tidak ada transaksi on-chain, tidak ada gas fee. Semua dihitung di Python.

Model yang dipakai sengaja **disederhanakan** untuk pembelajaran (bukan nasihat keuangan dan bukan prediksi harga).

```python
# 🧑‍💻 Parameter simulasi (boleh diubah untuk eksperimen Anda)
modal_simulasi = 1000      # $1,000 VIRTUAL — bukan uang nyata
hari = 365

apr_lending  = 0.05        # bunga lending 5%/tahun (asumsi)
apr_staking  = 0.04        # reward staking 4%/tahun (asumsi)
fee_tier     = 0.003       # fee swap pool 0,3%
volume_harian_pool = 1_000_000   # volume swap harian pool (USD virtual)
tvl_pool     = 10_000_000        # total likuiditas pool (USD virtual)
apr_insentif = 0.30        # insentif yield farming 30%/tahun dalam TOKEN reward
perubahan_harga_token_reward = -0.60   # skenario: harga token reward turun 60% dalam setahun

def bunga_majemuk_harian(modal, apr, hari):
    return modal * (1 + apr / 365) ** hari

# 1) Lending
nilai_lending = bunga_majemuk_harian(modal_simulasi, apr_lending, hari)
# 2) Staking (reward dibayar dalam token yang sama; asumsi harga token tetap)
nilai_staking = bunga_majemuk_harian(modal_simulasi, apr_staking, hari)
# 3) LP fee (tanpa impermanent loss dulu — lihat Section 8)
apr_fee_lp = (volume_harian_pool * fee_tier * 365) / tvl_pool
nilai_lp_fee = modal_simulasi * (1 + apr_fee_lp)
# 4) Yield farming: insentif dibayar dalam token reward yang harganya bisa jatuh
token_reward_usd_awal = modal_simulasi * apr_insentif
token_reward_usd_akhir = token_reward_usd_awal * (1 + perubahan_harga_token_reward)
nilai_farming = modal_simulasi + token_reward_usd_akhir

hasil_sim = pd.DataFrame({
    "Strategi (virtual)": ["Lending", "Staking", "Liquidity providing (fee saja)", "Yield farming (insentif token)"],
    "APR asumsi": [apr_lending, apr_staking, apr_fee_lp, apr_insentif],
    "Nilai akhir ($ virtual)": [nilai_lending, nilai_staking, nilai_lp_fee, nilai_farming],
})
hasil_sim["Return ($)"] = hasil_sim["Nilai akhir ($ virtual)"] - modal_simulasi
hasil_sim["Return (%)"] = hasil_sim["Return ($)"] / modal_simulasi * 100
display(hasil_sim.style.format({"APR asumsi": "{:.1%}", "Nilai akhir ($ virtual)": "{:,.2f}",
                                "Return ($)": "{:,.2f}", "Return (%)": "{:.2f}%"}))

ax = hasil_sim.plot.bar(x="Strategi (virtual)", y="Return ($)", legend=False, figsize=(9, 4),
                        color=["#2e86c1", "#27ae60", "#d68910", "#c0392b"])
ax.set_ylabel("Return 1 tahun ($ VIRTUAL)"); ax.set_title("Simulasi return — modal virtual $1,000")
plt.xticks(rotation=15, ha="right"); plt.tight_layout(); plt.show()
print(f"Catatan: APR yield farming tampak {apr_insentif:.0%}, tetapi karena token reward turun "
      f"{abs(perubahan_harga_token_reward):.0%}, return nyatanya hanya {token_reward_usd_akhir/modal_simulasi:.1%}.")
```

```python
# Reward simulation: kepekaan return yield farming terhadap harga token reward
skenario_harga = np.array([-0.9, -0.6, -0.3, 0.0, 0.3])
return_farming = (modal_simulasi * apr_insentif * (1 + skenario_harga)) / modal_simulasi * 100
tabel_reward = pd.DataFrame({"Perubahan harga token reward": [f"{s:+.0%}" for s in skenario_harga],
                             "Return yield farming (%)": return_farming.round(2)})
display(tabel_reward)
```

## SECTION 7 — APR VS APY · `EXPERIMENT`

- **APR** (*Annual Percentage Rate*): bunga tahunan **tanpa** pemajemukan.
- **APY** (*Annual Percentage Yield*): hasil tahunan **dengan** pemajemukan.

$$APY = \left(1 + \frac{APR}{n}\right)^{n} - 1$$

dengan $n$ = jumlah pemajemukan per tahun (1 = tahunan, 12 = bulanan, 365 = harian).

```python
def calculate_apy(apr, compounding):
    """Konversi APR (desimal, mis. 0.10) ke APY untuk `compounding` kali pemajemukan per tahun."""
    if compounding <= 0:
        raise ValueError("compounding harus > 0")
    return (1 + apr / compounding) ** compounding - 1

# Uji fungsi
assert abs(calculate_apy(0.10, 1) - 0.10) < 1e-12
print("APR 10%, harian →", f"{calculate_apy(0.10, 365):.4%}")

# Eksperimen: APR 5%, 10%, 20% × frekuensi pemajemukan
aprs = [0.05, 0.10, 0.20]
freq = {"Tahunan (1)": 1, "Bulanan (12)": 12, "Mingguan (52)": 52, "Harian (365)": 365}
tabel_apy = pd.DataFrame({nama: [calculate_apy(a, n) for a in aprs] for nama, n in freq.items()},
                         index=[f"APR {a:.0%}" for a in aprs])
display(tabel_apy.style.format("{:.3%}"))

fig, ax = plt.subplots(figsize=(8, 4))
ns = np.arange(1, 366)
for a in aprs:
    ax.plot(ns, [calculate_apy(a, n) * 100 for n in ns], label=f"APR {a:.0%}")
ax.set_xscale("log"); ax.set_xlabel("Jumlah pemajemukan per tahun (log)"); ax.set_ylabel("APY (%)")
ax.set_title("APY naik seiring frekuensi pemajemukan"); ax.legend(); ax.grid(alpha=0.3)
plt.tight_layout(); plt.show()
```

### ✍️ 7.1 Interpretasi

1. Untuk APR 20% dengan pemajemukan harian, APY = \((1+0{,}20/365)^{365}-1\) ≈ **22,13%**, sehingga selisihnya sekitar **2,13 poin persentase**. Selisih membesar pada APR lebih tinggi karena bunga yang telah diperoleh ikut menghasilkan bunga.
2. **APY tinggi tidak berarti risiko rendah**, karena (a) hasil bisa bersumber dari emisi token yang tidak berkelanjutan, (b) harga token reward dapat jatuh, (c) smart contract dapat memiliki bug/eksploit, dan (d) likuiditas, slashing, atau kondisi pasar dapat berubah.
3. Platform sering menampilkan APY karena pemajemukan membuat angka tahunan tampak lebih besar dan memudahkan perbandingan hasil dengan asumsi reward terus diinvestasikan kembali. Asumsi tersebut perlu diperiksa.
4. Pemajemukan harian di jaringan nyata mungkin membutuhkan transaksi dan gas fee. Jika biaya transaksi melebihi tambahan hasil, APY bersih menjadi lebih rendah; pada modal kecil, biaya bisa menghapus keuntungan.


## SECTION 8 — IMPERMANENT LOSS LAB · `EXPERIMENT → ANALYZE`

Pada AMM `x · y = k` dengan bobot 50:50, jika harga aset A berubah dengan rasio $r = P_{akhir}/P_{awal}$, maka nilai posisi LP relatif terhadap HODL:

$$IL(r) = \frac{2\sqrt{r}}{1 + r} - 1$$

**HODL** = simpan token apa adanya. **LP** = setor ke pool (dapat fee, tetapi terkena IL).

```python
def impermanent_loss(r):
    """Impermanent loss (desimal negatif) untuk rasio perubahan harga r pada pool 50:50."""
    return 2 * np.sqrt(r) / (1 + r) - 1

# Kondisi awal (virtual): $1,000 = $500 token A (ETH-like) + $500 stablecoin
modal_lp = modal_simulasi
harga_awal_A = 2000.0
jumlah_A = (modal_lp / 2) / harga_awal_A
jumlah_stable = modal_lp / 2
apr_fee_skenario = apr_fee_lp          # dari Section 6
periode_tahun = 1.0

rasio = [0.25, 0.5, 0.75, 1.0, 1.25, 1.5, 2.0, 3.0, 4.0]
rows = []
for r in rasio:
    harga_akhir = harga_awal_A * r
    nilai_hodl = jumlah_A * harga_akhir + jumlah_stable
    il = impermanent_loss(r)
    nilai_lp_tanpa_fee = nilai_hodl * (1 + il)
    fee = modal_lp * apr_fee_skenario * periode_tahun
    nilai_lp_dengan_fee = nilai_lp_tanpa_fee + fee
    rows.append([f"{r:.2f}x", harga_akhir, nilai_hodl, il * 100, nilai_lp_tanpa_fee,
                 nilai_lp_dengan_fee, nilai_lp_dengan_fee - nilai_hodl])
il_df = pd.DataFrame(rows, columns=["Rasio harga", "Harga akhir A", "HODL ($)", "IL (%)",
                                    "LP tanpa fee ($)", "LP + fee ($)", "LP+fee − HODL ($)"])
il_df["Pemenang"] = np.where(il_df["LP+fee − HODL ($)"] >= 0, "LP", "HODL")
display(il_df.style.format({"Harga akhir A": "{:,.0f}", "HODL ($)": "{:,.2f}", "IL (%)": "{:.2f}",
                            "LP tanpa fee ($)": "{:,.2f}", "LP + fee ($)": "{:,.2f}",
                            "LP+fee − HODL ($)": "{:+,.2f}"}))

fig, axes = plt.subplots(1, 2, figsize=(13, 4.5))
rr = np.linspace(0.1, 5, 300)
axes[0].plot(rr, impermanent_loss(rr) * 100, color="#c0392b")
axes[0].axhline(0, color="grey", lw=0.8); axes[0].axvline(1, color="grey", ls="--", lw=0.8)
axes[0].set_xlabel("Rasio harga akhir / awal"); axes[0].set_ylabel("Impermanent loss (%)")
axes[0].set_title("Kurva impermanent loss (pool 50:50)"); axes[0].grid(alpha=0.3)
x = np.arange(len(il_df))
axes[1].bar(x - 0.2, il_df["HODL ($)"], 0.4, label="HODL", color="#2e86c1")
axes[1].bar(x + 0.2, il_df["LP + fee ($)"], 0.4, label="LP + fee", color="#d68910")
axes[1].set_xticks(x); axes[1].set_xticklabels(il_df["Rasio harga"])
axes[1].set_ylabel("Nilai akhir ($ VIRTUAL)"); axes[1].set_title("HODL vs Liquidity Providing")
axes[1].legend(); axes[1].grid(axis="y", alpha=0.3)
plt.tight_layout(); plt.show()

# Titik impas: berapa APR fee minimal agar LP = HODL pada tiap rasio?
il_df["APR fee impas"] = [-impermanent_loss(r) * (jumlah_A * harga_awal_A * r + jumlah_stable) / modal_lp
                          for r in rasio]
print("APR fee minimum agar LP tidak kalah dari HODL:")
display(il_df[["Rasio harga", "IL (%)", "APR fee impas"]].style.format({"IL (%)": "{:.2f}", "APR fee impas": "{:.2%}"}))
```

### ✍️ 8.1 Interpretasi

1. Pada rasio harga selain 1, LP 50:50 umumnya mengalami **impermanent loss relatif terhadap HODL** sebelum fee. Contohnya, saat harga 2x IL sekitar −5,72% dan saat 4x sekitar −20,00%. Perubahan harga membuat komposisi pool menyesuaikan sehingga nilai LP dapat tertinggal dari memegang token awal.
2. Disebut *impermanent* karena kerugian relatif itu dapat mengecil jika rasio harga kembali ke titik awal sebelum posisi ditutup. Kerugian menjadi terealisasi ketika posisi ditarik pada kondisi yang merugikan; fee yang diperoleh dapat mengimbangi sebagian atau seluruh IL.
3. **USDC/USDT** biasanya memiliki IL lebih kecil karena kedua aset dirancang mengikuti nilai sekitar US$1, sehingga perubahan rasio harga cenderung kecil. Namun tetap ada risiko depeg, risiko penerbit/cadangan, smart contract, likuiditas, dan biaya.
4. Jika `apr_fee_skenario` diturunkan menjadi `0.02`, fee yang diterima lebih kecil sehingga lebih banyak skenario harga dapat membuat LP kalah dari HODL. Jalankan ulang sel Section 8 untuk melihat hasil tepatnya.

```python
STOP_FLAGS = {
    "minta_private_key_atau_seed": "Meminta private key / seed phrase",
    "wajib_deposit_atau_beli_token": "Wajib deposit / membeli token",
    "wajib_gas_aset_asli": "Wajib bayar gas fee dengan aset asli",
    "ada_biaya_pendaftaran": "Ada biaya pendaftaran/membership/voucher",
    "janji_hasil_pasti": "Menjanjikan hasil/keuntungan pasti",
    "stake_kripto_untuk_ikut": "Meminta stake kripto agar bisa ikut",
}
CAUTION_FLAGS = {
    "butuh_wallet": "Perlu alamat wallet (pakai wallet baru TANPA dana)",
    "butuh_kyc": "Perlu KYC / dokumen identitas",
    "sumber_tidak_resmi": "Info hanya dari media sosial / DM, bukan situs resmi",
    "status_tidak_jelas": "Tidak jelas masih aktif / deadline tidak tercantum",
    "minta_approve_kontrak": "Meminta connect wallet & approve kontrak",
}

def stop_checklist(nama, **flags):
    stop = [STOP_FLAGS[k] for k in STOP_FLAGS if flags.get(k)]
    caution = [CAUTION_FLAGS[k] for k in CAUTION_FLAGS if flags.get(k)]
    keputusan = "STOP" if stop else ("CAUTION" if caution else "GO")
    ikon = {"GO": "🟢", "CAUTION": "🟡", "STOP": "🔴"}[keputusan]
    print(f"{ikon} {nama}: {keputusan}")
    for s in stop: print("   ✖", s)
    for c in caution: print("   !", c)
    return keputusan

# 🧑‍💻 Contoh pemeriksaan — tambahkan pemeriksaan untuk peluang Anda sendiri
hasil_checklist = {
    "Superteam Earn bounty konten": stop_checklist("Superteam Earn bounty konten", butuh_wallet=True, butuh_kyc=True),
    "Code4rena (belajar laporan)": stop_checklist("Code4rena (belajar laporan)"),
    "ETHGlobal Mumbai": stop_checklist("ETHGlobal Mumbai", stake_kripto_untuk_ikut=True),
    "'Airdrop' dari DM Telegram": stop_checklist("'Airdrop' dari DM Telegram", sumber_tidak_resmi=True,
                                                minta_approve_kontrak=True, minta_private_key_atau_seed=True),
}
```
